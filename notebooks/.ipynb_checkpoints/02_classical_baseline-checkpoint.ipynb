```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 1. Load the data (Using relative path so it works on any drive)
col_names = [
    "duration","protocol_type","service","flag","src_bytes","dst_bytes",
    "land","wrong_fragment","urgent","hot","num_failed_logins","logged_in",
    "num_compromised","root_shell","su_attempted","num_root","num_file_creations",
    "num_shells","num_access_files","num_outbound_cmds","is_host_login",
    "is_guest_login","count","srv_count","serror_rate","srv_serror_rate",
    "rerror_rate","srv_rerror_rate","same_srv_rate","diff_srv_rate",
    "srv_diff_host_rate","dst_host_count","dst_host_srv_count",
    "dst_host_same_srv_rate","dst_host_diff_srv_rate","dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate","dst_host_serror_rate","dst_host_srv_serror_rate",
    "dst_host_rerror_rate","dst_host_srv_rerror_rate","label","difficulty"
]

train_df = pd.read_csv("../data/raw/KDDTrain+.txt", names=col_names)
test_df  = pd.read_csv("../data/raw/KDDTest+.txt", names=col_names)

# 2. Create binary labels (normal vs attack)
train_df['binary_label'] = train_df['label'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
test_df['binary_label']  = test_df['label'].apply(lambda x: 'normal' if x == 'normal' else 'attack')

# 3. Filter to ONLY normal traffic for training
train_normal = train_df[train_df['binary_label'] == 'normal']

# 4. Select the 11 numeric features and scale them
features = ['duration','src_bytes','dst_bytes','count','srv_count',
            'serror_rate','srv_serror_rate','rerror_rate','srv_rerror_rate',
            'same_srv_rate','diff_srv_rate']

scaler = MinMaxScaler()
X_train = scaler.fit_transform(train_normal[features])
X_test  = scaler.transform(test_df[features])
y_test  = test_df['binary_label'].values

print("✅ Data loaded and scaled successfully!")
print(f"Training shape (Normal only): {X_train.shape}")
print(f"Test shape (All traffic): {X_test.shape}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# 1. Check if we can use GPU, otherwise use CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 Using device: {device}")

# 2. Define the Autoencoder Architecture
class ClassicalAutoencoder(nn.Module):
    def __init__(self, input_dim):
        super(ClassicalAutoencoder, self).__init__()
        # Encoder: Compresses 11 features down to 8, then to 4 (the bottleneck)
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 8),
            nn.ReLU(),
            nn.Linear(8, 4)
        )
        # Decoder: Rebuilds the 4 features back out to 8, then to 11
        self.decoder = nn.Sequential(
            nn.Linear(4, 8),
            nn.ReLU(),
            nn.Linear(8, input_dim),
            nn.Sigmoid() # Sigmoid ensures output is between 0 and 1, matching our scaled data
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded

# 3. Initialize the model, loss function, and optimizer
input_dim = X_train.shape[1] # This is 11
model = ClassicalAutoencoder(input_dim).to(device)
criterion = nn.MSELoss() # We want to minimize the difference between input and output
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 4. Convert our numpy arrays to PyTorch tensors
X_train_tensor = torch.FloatTensor(X_train).to(device)

# 5. Train the model
print("\n🧠 Starting Training...")
epochs = 30
batch_size = 64

model.train()
for epoch in range(epochs):
    # Shuffle the data for better learning
    permutation = torch.randperm(X_train_tensor.size()[0])
    epoch_loss = 0
    
    # Train in mini-batches
    for i in range(0, X_train_tensor.size()[0], batch_size):
        indices = permutation[i:i+batch_size]
        batch_x = X_train_tensor[indices]
        
        optimizer.zero_grad()          # Clear old gradients
        outputs = model(batch_x)       # Forward pass
        loss = criterion(outputs, batch_x) # Calculate error
        loss.backward()                # Backward pass (calculate gradients)
        optimizer.step()               # Update weights
        epoch_loss += loss.item()
        
    avg_loss = epoch_loss / (X_train_tensor.size()[0] / batch_size)
    
    # Print progress every 10 epochs
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1}/{epochs}] | Loss: {avg_loss:.6f}")

print("✅ Training Complete!")
```

```python
import numpy as np

print("🔍 Evaluating model on test data to find the threshold...")

# 1. Set model to evaluation mode
model.eval()
X_test_tensor = torch.FloatTensor(X_test).to(device)

# 2. Pass the test data through the model (no gradient tracking needed)
with torch.no_grad():
    reconstructed = model(X_test_tensor)
    # Calculate the Mean Squared Error for EVERY single row
    mse_per_row = torch.mean((X_test_tensor - reconstructed) ** 2, dim=1).cpu().numpy()

# 3. Add these real error scores back to our test dataframe
test_df['reconstruction_error'] = mse_per_row

# 4. Calculate the Threshold
# We look ONLY at the errors from NORMAL traffic in the test set.
# We set the threshold at the 95th percentile. 
# Anything above this is considered an anomaly/attack.
normal_test_errors = test_df[test_df['binary_label'] == 'normal']['reconstruction_error']
THRESHOLD = np.percentile(normal_test_errors, 95)

# 5. Print the final results clearly
print("=" * 60)
print("🎯 FINAL RESULTS FOR DASHBOARD INTEGRATION")
print("=" * 60)
print(f"REAL THRESHOLD TO USE: {THRESHOLD:.6f}")
print("-" * 60)
print(f"Mean Error (Normal Traffic):  {normal_test_errors.mean():.6f}")
print(f"Mean Error (Attack Traffic):  {test_df[test_df['binary_label'] == 'attack']['reconstruction_error'].mean():.6f}")
print("=" * 60)

# Optional: Quick performance check
test_df['predicted'] = test_df['reconstruction_error'].apply(lambda x: 'attack' if x > THRESHOLD else 'normal')
accuracy = (test_df['binary_label'] == test_df['predicted']).mean() * 100
print(f"Baseline Detection Accuracy: {accuracy:.2f}%")
```
