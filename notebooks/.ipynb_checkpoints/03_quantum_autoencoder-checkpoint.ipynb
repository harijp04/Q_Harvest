```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 1. Load the data
col_names = [
    "duration","protocol_type","service","flag","src_bytes","dst_bytes",
    "land","wrong_fragment","urgent","hot","num_failed_logins","logged_in",
    "num_compromised","root_shell","su_attempted","num_root","num_file_creations",
    "num_shells","num_access_files","num_outbound_cmds","is_host_login",
    "is_guest_login","count","srv_count","serror_rate","srv_serror_rate",
    "rerror_rate","srv_rerror_rate","same_srv_rate","diff_srv_rate",
    "srv_diff_host_rate","dst_host_count","dst_host_srv_count",
    "dst_host_same_srv_rate","dst_host_diff_srv_rate","dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate","dst_host_serror_rate","dst_host_srv_serror_rate",
    "dst_host_rerror_rate","dst_host_srv_rerror_rate","label","difficulty"
]

train_df = pd.read_csv("../data/raw/KDDTrain+.txt", names=col_names)
test_df  = pd.read_csv("../data/raw/KDDTest+.txt", names=col_names)

# 2. Filter to ONLY normal traffic for training
train_df['binary_label'] = train_df['label'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
test_df['binary_label']  = test_df['label'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
train_normal = train_df[train_df['binary_label'] == 'normal']

# 3. Select features and scale them
features = ['duration','src_bytes','dst_bytes','count','srv_count',
            'serror_rate','srv_serror_rate','rerror_rate','srv_rerror_rate',
            'same_srv_rate','diff_srv_rate']

scaler = MinMaxScaler()
X_train = scaler.fit_transform(train_normal[features])
X_test  = scaler.transform(test_df[features])

print("✅ Data loaded and scaled in the new notebook!")
```

```python
import pennylane as qml
import torch
import torch.nn as nn

print("⚛️ Building the Quantum Autoencoder...")

# 1. Define the Quantum Device (Simulator)
# We use exactly 4 qubits to match our 4 PCA features!
n_qubits = 4
dev = qml.device("default.qubit", wires=n_qubits)

# 2. Define the Quantum Circuit (The QNode)
n_layers = 2  # Number of times we apply trainable quantum gates

@qml.qnode(dev, interface="torch")
def quantum_circuit(inputs, weights):
    # ENCODER: Map the 4 classical numbers into quantum states (rotating qubits)
    qml.AngleEmbedding(inputs, wires=range(n_qubits), rotation='Y')
    
    # PROCESSOR: Apply trainable quantum gates (this is what the AI "learns")
    qml.BasicEntanglerLayers(weights, wires=range(n_qubits))
    
    # DECODER: Measure the qubits to get 4 classical numbers back (outputs between -1 and 1)
    return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]

# 3. Define the shape of the trainable quantum weights
weight_shapes = {"weights": (n_layers, n_qubits)}

# 4. Wrap the quantum circuit into a PyTorch Layer
qlayer = qml.qnn.TorchLayer(quantum_circuit, weight_shapes)

# 5. Build the full Hybrid Quantum-Classical Model
class QuantumAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.qlayer = qlayer
        # Add a tiny classical layer at the end to scale the quantum output (-1 to 1) 
        # back to our original data scale (0 to 1)
        self.classical_output = nn.Linear(n_qubits, n_qubits)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        x = self.qlayer(x)               # Pass through quantum circuit
        x = self.classical_output(x)     # Scale dimensions
        x = self.sigmoid(x)              # Force output between 0 and 1
        return x

# 6. Initialize the model, loss, and optimizer
model_q = QuantumAutoencoder()
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model_q.parameters(), lr=0.01) # Slightly higher LR for quantum

print("✅ Quantum Autoencoder Built Successfully!")
print(f"Number of trainable quantum parameters: {sum(p.numel() for p in model_q.qlayer.parameters())}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Load the data
col_names = [
    "duration","protocol_type","service","flag","src_bytes","dst_bytes",
    "land","wrong_fragment","urgent","hot","num_failed_logins","logged_in",
    "num_compromised","root_shell","su_attempted","num_root","num_file_creations",
    "num_shells","num_access_files","num_outbound_cmds","is_host_login",
    "is_guest_login","count","srv_count","serror_rate","srv_serror_rate",
    "rerror_rate","srv_rerror_rate","same_srv_rate","diff_srv_rate",
    "srv_diff_host_rate","dst_host_count","dst_host_srv_count",
    "dst_host_same_srv_rate","dst_host_diff_srv_rate","dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate","dst_host_serror_rate","dst_host_srv_serror_rate",
    "dst_host_rerror_rate","dst_host_srv_rerror_rate","label","difficulty"
]

train_df = pd.read_csv("../data/raw/KDDTrain+.txt", names=col_names)
test_df  = pd.read_csv("../data/raw/KDDTest+.txt", names=col_names)

# Filter to ONLY normal traffic
train_df['binary_label'] = train_df['label'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
test_df['binary_label']  = test_df['label'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
train_normal = train_df[train_df['binary_label'] == 'normal']

# Select features and scale
features = ['duration','src_bytes','dst_bytes','count','srv_count',
            'serror_rate','srv_serror_rate','rerror_rate','srv_rerror_rate',
            'same_srv_rate','diff_srv_rate']

scaler = MinMaxScaler()
X_train = scaler.fit_transform(train_normal[features])
X_test  = scaler.transform(test_df[features])

print("✅ Data loaded and scaled!")
```

```python
from sklearn.decomposition import PCA

# Compress 11 features down to 4 for quantum
pca = PCA(n_components=4)
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

print(f"✅ PCA Complete! Shape: {X_train_pca.shape}")
```

```python
import torch

print("⚛️ Starting Quantum Training...")

# 1. Convert our PCA data to PyTorch tensors
X_train_tensor = torch.FloatTensor(X_train_pca)
X_test_tensor = torch.FloatTensor(X_test_pca)

# 2. Training settings
epochs = 15  # Quantum takes longer, so we do fewer epochs than classical
batch_size = 16 # Smaller batch size for quantum stability

model_q.train()
for epoch in range(epochs):
    permutation = torch.randperm(X_train_tensor.size()[0])
    epoch_loss = 0
    
    for i in range(0, X_train_tensor.size()[0], batch_size):
        indices = permutation[i:i+batch_size]
        batch_x = X_train_tensor[indices]
        
        optimizer.zero_grad()
        outputs = model_q(batch_x)
        loss = criterion(outputs, batch_x)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
        
    avg_loss = epoch_loss / (X_train_tensor.size()[0] / batch_size)
    print(f"Epoch [{epoch+1}/{epochs}] | Quantum Loss: {avg_loss:.6f}")

print("✅ Quantum Training Complete!")
```

```python
import numpy as np

print(" Evaluating Quantum model on test data...")

# 1. Set model to evaluation mode
model_q.eval()

# 2. Pass test data through the quantum model
with torch.no_grad():
    reconstructed = model_q(X_test_tensor)
    # Calculate Mean Squared Error for every row
    mse_per_row = torch.mean((X_test_tensor - reconstructed) ** 2, dim=1).numpy()

# 3. Add errors to dataframe
test_df['quantum_reconstruction_error'] = mse_per_row

# 4. Calculate the 95th percentile threshold for NORMAL traffic
normal_test_errors = test_df[test_df['binary_label'] == 'normal']['quantum_reconstruction_error']
QUANTUM_THRESHOLD = np.percentile(normal_test_errors, 95)

# 5. Print final results
print("=" * 60)
print("🎯 FINAL QUANTUM RESULTS")
print("=" * 60)
print(f"QUANTUM THRESHOLD: {QUANTUM_THRESHOLD:.6f}")
print("-" * 60)
print(f"Mean Error (Normal Traffic):  {normal_test_errors.mean():.6f}")
print(f"Mean Error (Attack Traffic):  {test_df[test_df['binary_label'] == 'attack']['quantum_reconstruction_error'].mean():.6f}")
print("=" * 60)
```

```python
import pandas as pd
import numpy as np
import torch

print("🚨 LIVE ATTACK SIMULATION 🚨")
print("=" * 70)

# 1. Get our threshold from the previous step
THRESHOLD = QUANTUM_THRESHOLD 

# 2. Pick 2 random NORMAL connections and 2 random ATTACK connections from test data
normal_samples = test_df[test_df['binary_label'] == 'normal'].sample(2, random_state=42)
attack_samples = test_df[test_df['binary_label'] == 'attack'].sample(2, random_state=42)

# Combine them for testing
test_samples = pd.concat([normal_samples, attack_samples])

# 3. Preprocess: Scale and apply PCA (exactly as the model expects)
sample_features = test_samples[features]
sample_scaled = scaler.transform(sample_features)
sample_pca = pca.transform(sample_scaled)
sample_tensor = torch.FloatTensor(sample_pca)

# 4. Pass through the Quantum Model
model_q.eval()
with torch.no_grad():
    reconstructed = model_q(sample_tensor)
    errors = torch.mean((sample_tensor - reconstructed) ** 2, dim=1).numpy()

# 5. Display the Results clearly
test_samples = test_samples.copy()
test_samples['quantum_error'] = errors
test_samples['verdict'] = test_samples['quantum_error'].apply(
    lambda x: "🚨 ATTACK DETECTED" if x > THRESHOLD else "✅ SAFE (Normal)"
)

# Format for clean printing
display_cols = ['binary_label', 'duration', 'src_bytes', 'dst_bytes', 'count', 'quantum_error', 'verdict']
print(test_samples[display_cols].to_string(index=False))
print("=" * 70)
print(f"Threshold used: {THRESHOLD:.6f}")
print("Notice how the 'attack' rows have errors MUCH higher than the threshold!")
```
